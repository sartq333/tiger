```python
import os
import pandas as pd
```

```python
movie_data = "../data/ml-1m"
dat_files = [os.path.join(movie_data, f) for f in os.listdir(movie_data) if f.endswith(".dat")]
print(dat_files)
```

```python
COLUMN_NAMES = {
    "movies.dat":  ["movie_id", "title", "genres"],
    "ratings.dat": ["user_id", "movie_id", "rating", "timestamp"],
    "users.dat":   ["user_id", "gender", "age", "occupation", "zipcode"],
}

class ExploreDAT:
    
    def __init__(self, dat_files):
        self.dfs = []
        for dat_file in dat_files:
            fname = os.path.basename(dat_file)
            df = pd.read_csv(dat_file, sep="::", engine="python", header=None,
                              encoding="latin-1", names=COLUMN_NAMES.get(fname))
            self.dfs.append({"dat_file": dat_file, "df": df})
    
    def shape_column_df(self):
        for dataframe in self.dfs:
            print("filename: ", dataframe["dat_file"],
                  "shape: ", dataframe["df"].shape,
                  "columns: ", dataframe["df"].columns.tolist())
            print(dataframe["df"].head())
            print(50*"-")
```

```python
explore_dat = ExploreDAT(dat_files=dat_files)
explore_dat.shape_column_df()
```
